```python
print("Q16")

df = pd.read_csv("/content/drive/MyDrive/DSAB13/student_performance.csv")

print("Missing Values:")
print(df.isnull().sum())

numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].mean()
)

df["Average_Score"] = (
    df["Python_Score"] +
    df["SQL_Score"] +
    df["Pandas_Score"] +
    df["NumPy_Score"]
) / 4

df["Pass_Status"] = np.where(
    df["Average_Score"] >= 50,
    "Pass",
    "Fail"
)


top5 = df.sort_values(
    by="Average_Score",
    ascending=False
).head(5)

print("\nTop 5 Students:")
print(top5[["Name", "Average_Score"]])


department_avg = df.groupby("Department")["Average_Score"].mean()

print("\nDepartment with Highest Average Score:")
print(department_avg.idxmax())


avg_attendance = df.groupby("Department")["Attendance"].mean()

print("\nAverage Attendance by Department:")
print(avg_attendance)


pass_fail = df["Pass_Status"].value_counts()

print("\nPass/Fail Count:")
print(pass_fail)


summary = df.groupby("Department").agg(
    Total_Students=("Student_ID", "count"),
    Average_Score=("Average_Score", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
    Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
).reset_index()

print("\nDepartment-wise Summary:")
print(summary)
```

```python
print("Q18")

df = pd.read_csv("/content/drive/MyDrive/DSAB13/employee.csv")

if df["Salary"].dtype == object:
    df["Salary"] = df["Salary"].astype(str).str.replace(",", "", regex=True)
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Experience"] = pd.to_numeric(df["Experience"], errors="coerce")
df["Performance_Score"] = pd.to_numeric(df["Performance_Score"], errors="coerce")

salary = df["Salary"].to_numpy()
performance = df["Performance_Score"].to_numpy()

mean_salary = np.nanmean(salary)
median_salary = np.nanmedian(salary)
std_salary = np.nanstd(salary)
mean_performance = np.nanmean(performance)

print("Mean Salary:", mean_salary)
print("Median Salary:", median_salary)
print("Standard Deviation of Salary:", std_salary)
print("Mean Performance Score:", mean_performance)


filtered_employees = df[
    (df["Experience"] > 3) &
    (df["Performance_Score"] >= 80)
]

print("\nEmployees satisfying both conditions:")
print(filtered_employees)


department_analysis = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Average_Performance_Score=("Performance_Score", "mean"),
    Total_Projects_Completed=("Projects_Completed", "sum"),
    Number_of_Employees=("Employee_ID", "count")
)

print("\nDepartment-wise Analysis:")
print(department_analysis)


highest_department = department_analysis[
    "Average_Performance_Score"
].idxmax()

highest_score = department_analysis[
    "Average_Performance_Score"
].max()

print("\nDepartment with highest average performance score:")
print(highest_department)
print("Average Performance Score:", highest_score)


overall_avg_salary = np.nanmean(salary)
overall_avg_performance = np.nanmean(performance)

result = df[
    (df["Salary"] > overall_avg_salary) &
    (df["Performance_Score"] < overall_avg_performance)
]

print("\nEmployees with above-average salary but below-average performance:")
print(result[
    ["Employee_ID", "Department", "Salary", "Performance_Score"]
])
```
